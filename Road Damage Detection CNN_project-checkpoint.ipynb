Road Damage Detection(CNN)
------
**Goal** -
To develop a CNN model that automatically classifies road images as normal, pothole or cracked roads

3)Initialize the dataset path 
-----

```python
# Directory path
base_dir = r"C:\Users\Pranali\OneDrive\Desktop\Imarticus\Capstone Project 2\road_damage_dataset"
train_dir = os.path.join(base_dir, 'train')
test_dir = os.path.join(base_dir, 'test')

# Training Directory path
train_crack_dir = os.path.join(train_dir, 'crack')
train_normal_dir = os.path.join(train_dir,'normal')
train_pothole_dir = os.path.join(train_dir, 'pothole')
```

4)Visualise random images
----

```python
#user definrd function to read random images
ncols = 4
nrows = 1
num_images = 4

def view_random_image(target_dir):
    fig = plt.gcf()
    fig.set_size_inches(ncols*4, nrows*4)

    random_image = random.sample(os.listdir(target_dir), num_images) # fetches 4 random images

    for i, img_path in enumerate(random_image): #enumerate fetches the imges and loop displays it
        ax = plt.subplot(nrows, ncols, i+1)
        ax.axis('off')

        img = mpimg.imread(os.path.join(target_dir , img_path))
        plt.imshow(img)
        plt.title(os.path.basename(target_dir))

    plt.show()
```

7)Image Data Preprocessing
----

```python
train_datagen = ImageDataGenerator(rescale=1/255.,
                                   rotation_range=15,
                                   width_shift_range=0.1,
                                   height_shift_range=0.1,
                                   shear_range=0.1,  
                                   zoom_range=0.1,
                                   horizontal_flip=True)

test_datagen = ImageDataGenerator(rescale=1/255.)
```

```python
train_generator = train_datagen.flow_from_directory(train_dir,
                                                   batch_size=32,
                                                   class_mode='categorical',
                                                   target_size=(224,224))

test_generator = train_datagen.flow_from_directory(test_dir,
                                                   batch_size=32,
                                                   class_mode='categorical',
                                                   target_size=(224,224),
                                                   shuffle=False)
```

8)Model Building 
----

```python
model = Sequential()
#convolution block 1
model.add(Conv2D(32,(3,3), activation='relu', input_shape=(224,224,3)))
model.add(MaxPooling2D((2,2)))
#convolution block 2
model.add(Conv2D(64,(3,3), activation='relu'))
model.add(MaxPooling2D(2,2))
#convolution block 3
model.add(Conv2D(128,(3,3), activation='relu'))
model.add(MaxPooling2D(2,2))

# Fully Connected Layers
model.add(Flatten())

model.add(Dense(128, activation='relu'))
model.add(Dropout(0.2))
#output layer
model.add(Dense(3, activation='softmax'))
```

```python
model.summary()
```

9)Model Compilation and Epochs Generation for training
----

```python
model.compile(loss='categorical_crossentropy',
             optimizer='adam',
             metrics=['accuracy'])

early_stopping = tf.keras.callbacks.EarlyStopping(patience=4, monitor='val_loss')

history = model.fit(train_generator,
                   epochs=30,
                   validation_data=test_generator,
                   callbacks=[early_stopping])
```

10)Plotting traing accuracy and loss curves
----

```python
def plot_loss_curves(history):
    loss = history.history['loss']
    val_loss = history.history['val_loss']

    accuracy = history.history['accuracy']
    val_accuracy = history.history['val_accuracy']

    epochs = range(len(history.history['loss']))

    # Plot loss
    plt.plot(epochs, loss, label="training_loss")
    plt.plot(epochs, val_loss, label="val_loss")
    plt.title('Loss')
    plt.xlabel('Epochs')
    plt.legend()

    # Plot accuracy
    plt.figure()
    plt.plot(epochs, accuracy, label="training_accuracy")
    plt.plot(epochs, val_accuracy, label="val_accuracy")
    plt.title('Accuracy')
    plt.xlabel('Epochs')
    plt.legend()
```

```python
plot_loss_curves(history)
```

11)Evaluate Loss and Accuracy
-----

```python
test_loss, test_accuracy = model.evaluate(test_generator)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
print("Test Accuracy:",test_accuracy * 100,"%")
```

```python
test_generator.reset()
predictions = model.predict(test_generator)
y_pred = np.argmax(predictions, axis=1)
y_true = test_generator.classes
```

```python
from sklearn.metrics import classification_report

class_names = list(test_generator.class_indices.keys())

print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names
    )
)
```
